```python
category_suffix = {
    "Beverages": "_BEV",
    "Butchers": "_BUT",
    "Computers and electric accessories": "_CEA",
    "Electric household essentials": "_EHE",
    "Food": "_FOOD",
    "Furniture": "_FUR",
    "Milk Products": "_MILK",
    "Patisserie": "_PAT"
}

expected_suffix = df["Category"].map(category_suffix)

actual_suffix = df["Item"].str.extract(r"(_[A-Z]+)$")[0]

item_mismatch = df[
    df["Item"].notna() &
    actual_suffix.ne(expected_suffix)
]

print("Number of category-item mismatches:", len(item_mismatch))
```

```python
df[["Price Per Unit", "Quantity", "Total Spent"]].describe()
```

```python
df["Calculated Total"] = (
    df["Price Per Unit"] * df["Quantity"]
)

df["Total Difference"] = (
    df["Total Spent"] - df["Calculated Total"]
)

comparison = df[
    ["Price Per Unit", "Quantity", "Total Spent",
     "Calculated Total", "Total Difference"]
]

comparison.head(10)
```

```python
complete_records = df[
    df["Price Per Unit"].notna() &
    df["Quantity"].notna() &
    df["Total Spent"].notna()
]

matching_records = complete_records[
    (complete_records["Total Spent"] -
     complete_records["Price Per Unit"] * complete_records["Quantity"]).abs() < 0.000001
]

mismatching_records = complete_records[
    (complete_records["Total Spent"] -
     complete_records["Price Per Unit"] * complete_records["Quantity"]).abs() >= 0.000001
]

print("Total records:", len(df))
print("Complete numerical records:", len(complete_records))
print("Matching records:", len(matching_records))
print("Mismatching records:", len(mismatching_records))
print("Records that cannot be checked:", len(df) - len(complete_records))
```

```python
numeric_columns = [
    "Price Per Unit",
    "Quantity",
    "Total Spent"
]

missing_patterns = (
    df[numeric_columns]
    .isna()
    .value_counts()
    .reset_index(name="Record Count")
)

missing_patterns
```

```python
price_missing_records = df[df["Price Per Unit"].isna()]

print("Records with missing price:", len(price_missing_records))

print(
    "Records with zero quantity:",
    (price_missing_records["Quantity"] == 0).sum()
)

print(
    "Records with missing quantity:",
    price_missing_records["Quantity"].isna().sum()
)

print(
    "Records with missing total:",
    price_missing_records["Total Spent"].isna().sum()
)
```

```python
calculated_prices = (
    price_missing_records["Total Spent"] /
    price_missing_records["Quantity"]
)

print("Calculated prices:")
print(calculated_prices.describe())

print("\nFirst 10 calculated prices:")
print(calculated_prices.head(10))
```

```python
df["Price Per Unit"] = df["Price Per Unit"].fillna(calculated_prices)
```

```python
print(
    "Remaining missing prices:",
    df["Price Per Unit"].isna().sum()
)
```

```python
discount_missing_summary = df.groupby(
    df["Discount Applied"],
    dropna=False
).size()

print(discount_missing_summary)
```

```python
numeric_columns = [
    "Price Per Unit",
    "Quantity",
    "Total Spent"
]

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    print("\nColumn:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
```

```python
total_q1 = df["Total Spent"].quantile(0.25)
total_q3 = df["Total Spent"].quantile(0.75)

total_iqr = total_q3 - total_q1

upper_limit = total_q3 + (1.5 * total_iqr)

potential_outliers = df[
    df["Total Spent"] > upper_limit
]

print("Upper IQR limit:", upper_limit)
print("Number of potential outliers:", len(potential_outliers))

potential_outliers[
    ["Transaction ID", "Category", "Item",
     "Price Per Unit", "Quantity", "Total Spent"]
]
```
